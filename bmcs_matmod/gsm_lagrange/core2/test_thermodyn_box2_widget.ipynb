# Test Widget - Debugging Duplicate Rendering

This notebook creates a minimal test case to debug the duplicate rendering issue.
The test widget mimics the structure of GSMThermodynBox2Widget but uses an actual
GSMThermodynBox2 instance and cycles through state functions F, G, H, U on each click.

```python
# Import the test widget and necessary components
from bmcs_matmod.gsm_lagrange.core2.test_thermodyn_box2_widget import TestWidget, create_test_widget
from bmcs_matmod.gsm_lagrange.core2.gsm_thermodyn_box import GSMThermodynBox
from bmcs_matmod.gsm_lagrange.core2.gsm_state_fn import GSMStateFn, StateFunction
from bmcs_matmod.gsm_lagrange.core2.gsm_vars import Scalar
import sympy as sp
import ipywidgets as widgets
from IPython.display import display, clear_output
```

## Create GSM Thermodynamic Box

First, let's create a simple GSMThermodynBox2 instance with a Helmholtz free energy function.

```python
# Create symbolic variables using enhanced Scalar class
T = Scalar(r'\vartheta', codename='T', real=True, positive=True)
S = Scalar('S', codename='S', real=True)
eps = Scalar(r'\varepsilon', codename='eps', real=True)
sig = Scalar(r'\sigma', codename='sig', real=True)
omega = Scalar(r'\omega', codename='omega', real=True, positive=True)
Y = Scalar('Y', codename='Y', real=True)

# Material parameters
E = Scalar('E', codename='E', positive=True)
C_eps = Scalar(r'C_{\varepsilon}', codename='C_eps', real=True, nonnegative=True)
T_0 = Scalar(r'\vartheta_0', codename='T_0', real=True, nonnegative=True)

# Create enhanced Helmholtz free energy expression F(T, ε, ω)
F_elastic = sp.Rational(1, 2) * (1 - omega) * E * eps**2
F_thermal = C_eps * (T - T * sp.log(T / T_0))
F_expr = F_elastic + F_thermal

# Create the initial state function instance
F_state_fn = GSMStateFn(
    fn_expr=F_expr,
    th_x_var=T,      # Temperature is natural variable
    th_y_var=S,      # Entropy is conjugate variable
    mc_x_var=eps,    # Strain is natural variable
    mc_y_var=sig,    # Stress is conjugate variable
    Eps_var=omega,   # Internal natural variable (damage)
    Sig_var=Y,       # Internal conjugate variable (stored energy)
    state_function_type=StateFunction.HELMHOLTZ
)

# Create the thermodynamic box
gsm_box = GSMThermodynBox(
    initial_state_fn=StateFunction.HELMHOLTZ,
    initial_state_instance=F_state_fn,
    T_var=T,
    S_var=S
)

print("GSM Thermodynamic Box created with Helmholtz free energy as initial state function.")
print(f"Expression: {F_expr}")
```

## Test 1: State Function Button Widget

This test uses the actual GSMThermodynBox2 instance with four individual
buttons (F, G, H, U) that directly display their corresponding state functions.

```python
# Create and show the test widget with GSMThermodynBox2 instance
test_widget = create_test_widget(gsm_box, "Test Widget - State Function Buttons")
test_widget.show()

print("Test widget created with four state function buttons.")
print("Click any button (F, G, H, U) to display its thermodynamic expression.")
print("Watch for duplicate rendering in debug messages.")
```
