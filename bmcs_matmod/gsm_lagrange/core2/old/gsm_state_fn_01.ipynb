# GSMStateFn: State Function Accessor and Cursor

This notebook demonstrates the `GSMStateFn` class, which serves as a dynamic cursor/accessor for navigating and computing with thermodynamic state functions within the framework provided by `GSMThermodynBox`.

## Architecture Overview

- **GSMThermodynBox**: Static framework providing the "playground" 
- **GSMStateFn**: Dynamic cursor for computational operations
- **StateFunctionTag**: Enum identifying state function types

The cursor pattern allows efficient navigation between different thermodynamic potentials while maintaining access to the complete framework.

```python
# Setup and imports
import sys
import os

# Add the core2 module to the path
current_dir = os.path.dirname(os.path.abspath('__file__' if '__file__' in globals() else 'gsm_state_fn_01.ipynb'))
sys.path.insert(0, current_dir)

from gsm_thermodyn_box import GSMThermodynBox, StateFunctionTag
from gsm_state_fn import GSMStateFn
import sympy as sp

print("Imports successful!")
print(f"Available state functions: {[tag.name for tag in StateFunctionTag]}")

# Create the thermodynamic framework
box = GSMThermodynBox()
print(f"Created thermodynamic box with {len(box.NATURAL_VARIABLES_MAPPING)} state functions")
```

## 1. Creating State Function Accessors

State function accessors are created from the thermodynamic box and can start with any state function as the cursor position.

```python
# Create accessors for different state functions
print("Creating state function accessors:")
print("=" * 40)

# Default accessor (starts with internal energy)
state_fn_U = box.create_state_fn()
print(f"U accessor: cursor at {state_fn_U.current_state_fn_tag.name}")

# Specific state function accessors
state_fn_F = box.create_state_fn(StateFunctionTag.HELMHOLTZ)
state_fn_G = box.create_state_fn(StateFunctionTag.GIBBS)
state_fn_H = box.create_state_fn(StateFunctionTag.ENTHALPY)

print(f"F accessor: cursor at {state_fn_F.current_state_fn_tag.name}")
print(f"G accessor: cursor at {state_fn_G.current_state_fn_tag.name}")
print(f"H accessor: cursor at {state_fn_H.current_state_fn_tag.name}")

# Each accessor provides the same interface but starts at different points
print(f"\nAll accessors have the same methods:")
methods = [m for m in dir(state_fn_U) if not m.startswith('_') and callable(getattr(state_fn_U, m))]
print(f"Available methods: {len(methods)}")
for method in sorted(methods)[:5]:
    print(f"  - {method}")
print("  ... and more")
```

## 2. Cursor Navigation

The state function accessors act as cursors that can navigate between different thermodynamic potentials.

```python
# Demonstrate cursor navigation
print("Cursor Navigation Demonstration:")
print("=" * 35)

# Start with internal energy
state_fn = box.create_state_fn(StateFunctionTag.INTERNAL_ENERGY)
print(f"Starting cursor position: {state_fn.current_state_fn_tag.name}")

# The accessor provides direct methods for each state function
print(f"\nDirect navigation methods:")
print(f"- state_fn.U() -> Navigate to Internal Energy")
print(f"- state_fn.F() -> Navigate to Helmholtz Free Energy")
print(f"- state_fn.H() -> Navigate to Enthalpy")
print(f"- state_fn.G() -> Navigate to Gibbs Free Energy")

# Navigate to different state functions
print(f"\nNavigation examples:")
current_pos = state_fn.current_state_fn_tag.name
print(f"Current: {current_pos}")

# Navigate to Helmholtz
state_fn_at_F = state_fn.F()
print(f"After state_fn.F(): {state_fn_at_F.current_state_fn_tag.name}")

# Navigate to Gibbs  
state_fn_at_G = state_fn_at_F.G()
print(f"After .G(): {state_fn_at_G.current_state_fn_tag.name}")

# Each navigation returns a new accessor positioned at the target
print(f"\nCursor pattern properties:")
print(f"- Original accessor still at: {state_fn.current_state_fn_tag.name}")
print(f"- F accessor at: {state_fn_at_F.current_state_fn_tag.name}")
print(f"- G accessor at: {state_fn_at_G.current_state_fn_tag.name}")
```

```python
# Create different state function accessors
print("Creating state function accessors:")
print("=" * 40)

# Default accessor (starts with U - internal energy)
state_fn_u = box.create_state_fn()
print(f"Default accessor: {state_fn_u.current_state_fn.name} ({state_fn_u.current_state_fn.value})")

# Start with different state functions
state_fn_f = box.create_state_fn(StateFunctionTag.F)  # Helmholtz free energy
state_fn_g = box.create_state_fn(StateFunctionTag.G)  # Gibbs free energy  
state_fn_h = box.create_state_fn(StateFunctionTag.H)  # Enthalpy

print(f"Helmholtz accessor: {state_fn_f.current_state_fn.name} ({state_fn_f.current_state_fn.value})")
print(f"Gibbs accessor: {state_fn_g.current_state_fn.name} ({state_fn_g.current_state_fn.value})")
print(f"Enthalpy accessor: {state_fn_h.current_state_fn.name} ({state_fn_h.current_state_fn.value})")

# All accessors share the same framework
print(f"\nAll accessors use the same framework:")
print(f"Same box reference: {state_fn_u.thermodynamic_box is state_fn_f.thermodynamic_box}")

# Show current natural variables
print(f"\nCurrent natural variables for each:")
for name, accessor in [("U", state_fn_u), ("F", state_fn_f), ("G", state_fn_g), ("H", state_fn_h)]:
    vars = accessor.get_current_natural_variables()
    print(f"  {name}: {[var.name for var in vars]}")
```
