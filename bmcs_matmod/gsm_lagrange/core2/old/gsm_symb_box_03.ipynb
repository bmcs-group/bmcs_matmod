```python
# Import and Setup with Fresh Module Loading
import sympy as sp
import sys
from pathlib import Path

# Clear any cached modules for completely fresh testing
modules_to_clear = ['gsm_symb_box']
for module in modules_to_clear:
    if module in sys.modules:
        del sys.modules[module]

# Add core2 to path (current directory)
sys.path.insert(0, str(Path.cwd()))

# Fresh imports
from gsm_symb_box import GSMSymbBox, StateFunction
from gsm_vars import Scalar  # Import customized symbol definitions
from IPython.display import display, Math

# Enable pretty printing
sp.init_printing()

print("GSM Thermodynamic Box - Comprehensive Verification")
print("=" * 60)
print("PRIMARY CRITERION: Round-trip consistency")
print("If F→G→F = F exactly, then transformations are mathematically correct")
print("✓ Using customized Scalar symbols with proper LaTeX rendering")
print("=" * 60)
```

```python
# Define Test Model and Initialize GSMSymbBox
print("Test Model: Elastic-Damage Material")
print("=" * 40)

# Define symbolic variables using customized Scalar class with proper LaTeX symbols and codenames
# Following the naming conventions from VARIABLE_NAMING.md

# Thermal variables (Category 0)
T = Scalar('T', codename='T', real=True, positive=True)  # Temperature (intensive)
S = Scalar('S', codename='S', real=True)                 # Entropy (extensive)

# External mechanical variables (Category 0.1)
eps = Scalar(r'\varepsilon', codename='eps', real=True)  # External strain (extensive)
sig = Scalar(r'\sigma', codename='sig', real=True)       # External stress (intensive)

# Damage variables (Category 2)
omega = Scalar(r'\omega', codename='omega', real=True, positive=True)  # Damage parameter
Y = Scalar('Y', codename='Y', real=True)                               # Damage driving force

# Material parameters
E = Scalar('E', codename='E', positive=True)  # Young's modulus

print("Symbolic variables defined with proper LaTeX notation:")
print(f"Temperature:     {T} (codename: {T.codename})")
print(f"Entropy:         {S} (codename: {S.codename})")  
print(f"Strain:          {eps} (codename: {eps.codename})")
print(f"Stress:          {sig} (codename: {sig.codename})")
print(f"Damage:          {omega} (codename: {omega.codename})")
print(f"Damage force:    {Y} (codename: {Y.codename})")
print(f"Young's modulus: {E} (codename: {E.codename})")

# Elastic-damage Helmholtz free energy
F_original = sp.Rational(1, 2) * (1 - omega) * E * eps**2

print(f"\nElastic-damage Helmholtz free energy:")
print(f"F(T,ε,ω) = {F_original}")
display(Math(f"F(T, \\varepsilon, \\omega) = {sp.latex(F_original)}"))

# Create GSMSymbBox using the customized symbols
# Note: initial_state_fn is now a required keyword argument
print("\nCreating GSMSymbBox...")
box = GSMSymbBox(
    initial_state_fn=StateFunction.HELMHOLTZ,  # Required keyword argument
    T_var=T,
    S_var=S,
    eps_vars=(eps,),
    sig_vars=(sig,),
    Eps_vars=(omega,),
    Sig_vars=(Y,),
    m_params=(E,),
    initial_expression=F_original
)

print(f"✓ GSMSymbBox created successfully with customized symbols")
print(f"✓ Initial state function: {box.current_state_fn.value}")
print(f"✓ Initial expression: {box.get_current_expression()}")
print(f"✓ All symbols follow VARIABLE_NAMING.md conventions")
print(f"✓ LaTeX rendering: ε, σ, ω will display with proper mathematical notation")
```

```python
# Test 1: Property-Based Access to All State Functions
print("TEST 1: Property-Based Access to All State Functions")
print("=" * 60)

# Access all state functions via properties
print("Accessing all state functions via properties...")

F_prop = box.F
G_prop = box.G
U_prop = box.U
H_prop = box.H

print(f"\nF (Helmholtz): {F_prop}")
print(f"G (Gibbs):     {G_prop}")
print(f"U (Internal):  {U_prop}")
print(f"H (Enthalpy):  {H_prop}")

# Display with LaTeX
print("\nFormatted expressions:")
display(Math(f"F(T,\\varepsilon,\\omega) = {sp.latex(F_prop)}"))
display(Math(f"G(T,\\sigma,\\omega) = {sp.latex(G_prop)}"))
display(Math(f"U(S,\\varepsilon,\\omega) = {sp.latex(U_prop)}"))
display(Math(f"H(S,\\sigma,\\omega) = {sp.latex(H_prop)}"))

print("\n✓ All four state functions accessible via properties")
```

```python
# Test 2: Round-Trip Consistency - Adjacent Transformations
print("TEST 2: Round-Trip Consistency - Adjacent Transformations")
print("=" * 65)

print("Testing all adjacent round-trips (direct transformations)...")

# Store original expressions
original_expressions = {
    'F': F_prop,
    'G': G_prop,
    'U': U_prop,
    'H': H_prop
}

# Test 2.1: F → G → F
print("\n2.1: F → G → F (Mechanical conjugate transformation)")
print("-" * 55)

box.set_current_state_function(StateFunction.HELMHOLTZ, F_prop)
G_from_F = box.legendre_transform(StateFunction.GIBBS)

box.set_current_state_function(StateFunction.GIBBS, G_from_F)
F_recovered = box.legendre_transform(StateFunction.HELMHOLTZ)

FGF_diff = sp.simplify(F_recovered - F_prop)
FGF_success = FGF_diff == 0

print(f"Original F:  {F_prop}")
print(f"G from F:    {G_from_F}")
print(f"F recovered: {F_recovered}")
print(f"Difference:  {FGF_diff}")
print(f"F→G→F: {'✓ PASS' if FGF_success else '✗ FAIL'}")

# Test 2.2: F → U → F  
print("\n2.2: F → U → F (Thermal conjugate transformation)")
print("-" * 52)

box.set_current_state_function(StateFunction.HELMHOLTZ, F_prop)
U_from_F = box.legendre_transform(StateFunction.INTERNAL_ENERGY)

box.set_current_state_function(StateFunction.INTERNAL_ENERGY, U_from_F)
F_from_U = box.legendre_transform(StateFunction.HELMHOLTZ)

FUF_diff = sp.simplify(F_from_U - F_prop)
FUF_success = FUF_diff == 0

print(f"Original F:  {F_prop}")
print(f"U from F:    {U_from_F}")
print(f"F from U:    {F_from_U}")
print(f"Difference:  {FUF_diff}")
print(f"F→U→F: {'✓ PASS' if FUF_success else '✗ FAIL'}")

# Test 2.3: G → H → G
print("\n2.3: G → H → G (Thermal conjugate transformation)")
print("-" * 52)

box.set_current_state_function(StateFunction.GIBBS, G_prop)
H_from_G = box.legendre_transform(StateFunction.ENTHALPY)

box.set_current_state_function(StateFunction.ENTHALPY, H_from_G)
G_from_H = box.legendre_transform(StateFunction.GIBBS)

GHG_diff = sp.simplify(G_from_H - G_prop)
GHG_success = GHG_diff == 0

print(f"Original G:  {G_prop}")
print(f"H from G:    {H_from_G}")
print(f"G from H:    {G_from_H}")
print(f"Difference:  {GHG_diff}")
print(f"G→H→G: {'✓ PASS' if GHG_success else '✗ FAIL'}")

# Summary
adjacent_tests = [FGF_success, FUF_success, GHG_success]
adjacent_success = all(adjacent_tests)

print("\n" + "=" * 40)
print(f"ADJACENT ROUND-TRIPS: {'✓ ALL PASS' if adjacent_success else '✗ SOME FAIL'}")
print(f"F→G→F: {'✓' if FGF_success else '✗'}")
print(f"F→U→F: {'✓' if FUF_success else '✗'}")
print(f"G→H→G: {'✓' if GHG_success else '✗'}")
print("=" * 40)
```

```python
# Test 3: Round-Trip Consistency - Diagonal Transformations
print("TEST 3: Round-Trip Consistency - Diagonal Transformations")
print("=" * 65)

print("Testing diagonal round-trips (two-step transformations)...")

# Test 3.1: F → G → F (via diagonal path F→G directly)
print("\n3.1: F → G → F (already tested above)")
print(f"Result: {'✓ PASS' if FGF_success else '✗ FAIL'}")

# Test 3.2: U → G → U (diagonal transformation)
print("\n3.2: U → G → U (Diagonal transformation)")
print("-" * 43)

box.set_current_state_function(StateFunction.INTERNAL_ENERGY, U_prop)
G_from_U = box.legendre_transform(StateFunction.GIBBS)

box.set_current_state_function(StateFunction.GIBBS, G_from_U)
U_from_G = box.legendre_transform(StateFunction.INTERNAL_ENERGY)

UGU_diff = sp.simplify(U_from_G - U_prop)
UGU_success = UGU_diff == 0

print(f"Original U:  {U_prop}")
print(f"G from U:    {G_from_U}")
print(f"U from G:    {U_from_G}")
print(f"Difference:  {UGU_diff}")
print(f"U→G→U: {'✓ PASS' if UGU_success else '✗ FAIL'}")

# Test 3.3: H → F → H (diagonal transformation)
print("\n3.3: H → F → H (Diagonal transformation)")
print("-" * 43)

box.set_current_state_function(StateFunction.ENTHALPY, H_prop)
F_from_H = box.legendre_transform(StateFunction.HELMHOLTZ)

box.set_current_state_function(StateFunction.HELMHOLTZ, F_from_H)
H_from_F = box.legendre_transform(StateFunction.ENTHALPY)

HFH_diff = sp.simplify(H_from_F - H_prop)
HFH_success = HFH_diff == 0

print(f"Original H:  {H_prop}")
print(f"F from H:    {F_from_H}")
print(f"H from F:    {H_from_F}")
print(f"Difference:  {HFH_diff}")
print(f"H→F→H: {'✓ PASS' if HFH_success else '✗ FAIL'}")

# Summary
diagonal_tests = [UGU_success, HFH_success]
diagonal_success = all(diagonal_tests)

print("\n" + "=" * 40)
print(f"DIAGONAL ROUND-TRIPS: {'✓ ALL PASS' if diagonal_success else '✗ SOME FAIL'}")
print(f"U→G→U: {'✓' if UGU_success else '✗'}")
print(f"H→F→H: {'✓' if HFH_success else '✗'}")
print("=" * 40)
```

```python
# Test 4: Property Access Consistency
print("TEST 4: Property Access Consistency")
print("=" * 40)

print("Testing consistency between property access and explicit transformations...")

# Reset to clean state
box.set_current_state_function(StateFunction.HELMHOLTZ, F_original)
box.explicit_expressions = {StateFunction.HELMHOLTZ: F_original}

# Get via properties
F_via_prop = box.F
G_via_prop = box.G
U_via_prop = box.U
H_via_prop = box.H

# Get via explicit transformations
box.set_current_state_function(StateFunction.HELMHOLTZ, F_original)
G_via_explicit = box.legendre_transform(StateFunction.GIBBS)
U_via_explicit = box.legendre_transform(StateFunction.INTERNAL_ENERGY)
H_via_explicit = box.legendre_transform(StateFunction.ENTHALPY)
F_via_explicit = box.legendre_transform(StateFunction.HELMHOLTZ)

# Compare
F_consistent = sp.simplify(F_via_prop - F_via_explicit) == 0
G_consistent = sp.simplify(G_via_prop - G_via_explicit) == 0
U_consistent = sp.simplify(U_via_prop - U_via_explicit) == 0
H_consistent = sp.simplify(H_via_prop - H_via_explicit) == 0

print(f"\nProperty vs Explicit Comparison:")
print(f"F consistent: {'✓' if F_consistent else '✗'}")
print(f"G consistent: {'✓' if G_consistent else '✗'}")
print(f"U consistent: {'✓' if U_consistent else '✗'}")
print(f"H consistent: {'✓' if H_consistent else '✗'}")

property_consistency = all([F_consistent, G_consistent, U_consistent, H_consistent])

print("\n" + "=" * 30)
print(f"PROPERTY CONSISTENCY: {'✓ PASS' if property_consistency else '✗ FAIL'}")
print("=" * 30)
```

```python
# Test 5: Constitutive Relations from Final Expressions
print("TEST 5: Constitutive Relations from Final State Functions")
print("=" * 65)

print("Extracting constitutive relations from the final state function expressions...")
print("Note: These should be consistent across transformations.")

# Use the consistent expressions from property access
F_final = box.F
G_final = box.G
U_final = box.U
H_final = box.H

print(f"\nFinal expressions:")
print(f"F = {F_final}")
print(f"G = {G_final}")
print(f"U = {U_final}")
print(f"H = {H_final}")

# Extract constitutive relations
print("\n5.1: Stress-Strain Relations")
print("-" * 30)

sigma_from_F = sp.diff(F_final, eps)  # σ = ∂F/∂ε
epsilon_from_G = -sp.diff(G_final, sig)  # ε = -∂G/∂σ
sigma_from_U = sp.diff(U_final, eps)  # σ = ∂U/∂ε
epsilon_from_H = -sp.diff(H_final, sig)  # ε = -∂H/∂σ

print(f"σ from F: {sigma_from_F}")
print(f"ε from G: {epsilon_from_G}")
print(f"σ from U: {sigma_from_U}")
print(f"ε from H: {epsilon_from_H}")

# Check stress consistency
stress_FU_consistent = sp.simplify(sigma_from_F - sigma_from_U) == 0
print(f"\nStress consistency (F vs U): {'✓' if stress_FU_consistent else '✗'}")

# Check strain consistency
strain_GH_consistent = sp.simplify(epsilon_from_G - epsilon_from_H) == 0
print(f"Strain consistency (G vs H): {'✓' if strain_GH_consistent else '✗'}")

print("\n5.2: Damage Force Relations")
print("-" * 28)

Y_from_F = -sp.diff(F_final, omega)  # Y = -∂F/∂ω
Y_from_G = -sp.diff(G_final, omega)  # Y = -∂G/∂ω
Y_from_U = -sp.diff(U_final, omega)  # Y = -∂U/∂ω
Y_from_H = -sp.diff(H_final, omega)  # Y = -∂H/∂ω

print(f"Y from F: {Y_from_F}")
print(f"Y from G: {Y_from_G}")
print(f"Y from U: {Y_from_U}")
print(f"Y from H: {Y_from_H}")

# Check damage force consistency (should be the same after variable substitution)
# For proper comparison, substitute constitutive relations to express all in same variables

print("\n5.3: Inverse Relationship Test")
print("-" * 32)

# Test if σ(ε) and ε(σ) are inverses
epsilon_test = epsilon_from_G.subs(sig, sigma_from_F)
epsilon_simplified = sp.simplify(epsilon_test)
inverse_correct = sp.simplify(epsilon_simplified - eps) == 0

print(f"σ(ε) = {sigma_from_F}")
print(f"ε(σ) = {epsilon_from_G}")
print(f"ε(σ(ε)) = {epsilon_simplified}")
print(f"Should equal ε = {eps}")
print(f"Inverse relationship: {'✓ CORRECT' if inverse_correct else '✗ INCORRECT'}")

if not inverse_correct:
    difference = sp.simplify(epsilon_simplified - eps)
    print(f"Difference: {difference}")

# Summary
constitutive_tests = [stress_FU_consistent, strain_GH_consistent, inverse_correct]
constitutive_success = all(constitutive_tests)

print("\n" + "=" * 40)
print(f"CONSTITUTIVE RELATIONS: {'✓ ALL CONSISTENT' if constitutive_success else '⚠ NEED REVIEW'}")
print(f"Stress consistency: {'✓' if stress_FU_consistent else '✗'}")
print(f"Strain consistency: {'✓' if strain_GH_consistent else '✗'}")
print(f"Inverse relations: {'✓' if inverse_correct else '✗'}")
print("=" * 40)
```
