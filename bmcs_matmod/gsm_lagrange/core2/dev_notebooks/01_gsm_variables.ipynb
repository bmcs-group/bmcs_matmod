### Understanding the Scalar Interface

The `Scalar` class provides a dual-name system:
- **`name`**: Used for SymPy operations and LaTeX rendering (can contain LaTeX code)
- **`codename`**: Used for code generation, lambdification, and numerical mapping

**Interface**: `Scalar(name, codename=None)`

**Usage Patterns**:
1. **Simple variables**: `Scalar('eps')` → name='eps', codename='eps'
2. **LaTeX display**: `Scalar(r'\varepsilon', 'eps')` → LaTeX ε, codename 'eps'  
3. **Descriptive codenames**: `Scalar('E', 'elastic_modulus')` → name='E', codename='elastic_modulus'

This enables mathematical clarity through LaTeX while maintaining simple codenames for numerical computation.

### Scalar variables

```python
# Create example variables with descriptions and assumptions
variables = [
    Scalar(r'\varepsilon', 'eps', 'Engineering strain', real=True, finite=True),
    Scalar(r'\sigma', 'sig', 'Cauchy stress', real=True, finite=True),
    Scalar('T', 'temperature', 'Temperature variable', real=True, positive=True),
    Scalar(r'\omega', 'omega', 'Damage parameter', real=True, nonnegative=True),
]

# Import the table creation function and Markdown display
from bmcs_matmod.gsm_lagrange.core2.gsm_vars import markdown_vars_table
from IPython.display import Markdown

# Generate and display the table
table_markdown = markdown_vars_table(variables)
display(Markdown(table_markdown))
```

### Material Parameter Variables

```python
# Create material parameter variables
material_params = [
    Scalar('E', 'E', 'Elastic modulus', positive=True, real=True),
    Scalar(r'\nu', 'poisson_ratio', 'Poisson ratio', real=True),
    Scalar(r'\alpha', 'thermal_expansion', 'Thermal expansion coefficient', real=True),
    Scalar(r'\kappa', 'bulk_modulus', 'Bulk modulus', positive=True, real=True),
    Scalar(r'\mu', 'shear_modulus', 'Shear modulus', positive=True, real=True)
]

# Display material parameters table
table_markdown = markdown_vars_table(material_params)
display(Markdown(table_markdown))
```
