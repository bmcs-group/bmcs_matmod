## Create GSM Symbolic Box

Let's create a GSM symbolic box with thermodynamic state functions for demonstration.

````python
# Define Extended Test Model and Initialize GSMSymbBox
print("Extended Test Model: Elastic-Damage Material with Thermal Capacity")
print("=" * 70)

# Define symbolic variables using customized Scalar class with proper LaTeX symbols and codenames
# Following the naming conventions from VARIABLE_NAMING.md

# Thermal variables (Category 0)
T = Scalar(r'\vartheta', codename='T', real=True, positive=True)  # Temperature (intensive)
S = Scalar('S', codename='S', real=True)                 # Entropy (extensive)

# External mechanical variables (Category 0.1)
eps = Scalar(r'\varepsilon', codename='eps', real=True)  # External strain (extensive)
sig = Scalar(r'\sigma', codename='sig', real=True)       # External stress (intensive)

# Damage variables (Category 2)
omega = Scalar(r'\omega', codename='omega', real=True, positive=True)  # Damage parameter
Y = Scalar('Y', codename='Y', real=True)                               # Damage driving force

# Material parameters
E = Scalar('E', codename='E', positive=True)  # Young's modulus

# NEW: Thermal capacity parameters
C_v = Scalar(r'C_{\mathrm{v}}', codename='C_v', real=True, nonnegative=True)  # Thermal capacity
T_0 = Scalar(r'\vartheta_0', codename='T_0', real=True, nonnegative=True)     # Reference temperature

print("Symbolic variables defined with proper LaTeX notation:")
print(f"Temperature:      {T} (codename: {T.codename})")
print(f"Entropy:          {S} (codename: {S.codename})")  
print(f"Strain:           {eps} (codename: {eps.codename})")
print(f"Stress:           {sig} (codename: {sig.codename})")
print(f"Damage:           {omega} (codename: {omega.codename})")
print(f"Damage force:     {Y} (codename: {Y.codename})")
print(f"Young's modulus:  {E} (codename: {E.codename})")
print(f"Thermal capacity: {C_v} (codename: {C_v.codename})")
print(f"Reference temp:   {T_0} (codename: {T_0.codename})")

# Extended Helmholtz free energy: elastic-damage + thermal capacity
F_elastic = sp.Rational(1, 2) * (1 - omega) * E * eps**2
F_thermal = -C_v * (T - T_0)**2 / (2 * T_0)
F_original = F_elastic + F_thermal

print(f"\nExtended Helmholtz free energy components:")
print(f"F_elastic = {F_elastic}")
print(f"F_thermal = {F_thermal}")
print(f"F_total(T,ε,ω) = {F_original}")

display(Math(f"F_{{\\text{{elastic}}}} = {sp.latex(F_elastic)}"))
display(Math(f"F_{{\\text{{thermal}}}} = {sp.latex(F_thermal)}"))
display(Math(f"F_{{\\text{{total}}}}(T, \\varepsilon, \\omega) = {sp.latex(F_original)}"))

# Create GSMSymbBox using the customized symbols
# Note: initial_state_fn is now a required keyword argument
print("\nCreating GSMSymbBox with extended thermal model...")
box = GSMSymbBox(
    initial_state_fn=StateFunction.HELMHOLTZ,  # Required keyword argument
    T_var=T,
    S_var=S,
    eps_vars=(eps,),
    sig_vars=(sig,),
    Eps_vars=(omega,),
    Sig_vars=(Y,),
    m_params=(E, C_v, T_0),  # Include thermal parameters
    initial_expression=F_original
)

print(f"✓ GSMSymbBox created successfully with extended thermal model")
print(f"✓ Initial state function: {box.current_state_fn.value}")
print(f"✓ Initial expression: {box.get_current_expression()}")
print(f"✓ All symbols follow VARIABLE_NAMING.md conventions")
print(f"✓ LaTeX rendering: ε, σ, ω, Cᵥ, ϑ₀ will display with proper mathematical notation")
````
